## 2. Loading the Dataset

**Purpose:** Locate and load the original dataset.

`find_input()` locates the CSV or Excel file.
`load_data()` reads the file into a Pandas DataFrame.

A DataFrame is a two-dimensional table containing rows and columns.

```python
dataset_path = an.find_input()
raw = an.load_data(dataset_path)

print('Dataset Name:', dataset_path.name)
print('Number of Rows:', raw.shape[0])
print('Number of Columns:', raw.shape[1])

display(raw.head())
```

## 3. Understanding the Dataset

**Purpose:** Understand the structure, column names and data types.

- `head()` displays the first five records.
- `info()` displays column names and data types.

## 4. Data Cleaning and Preprocessing

**Purpose:** Improve data quality before performing analysis.

The cleaning function performs the following operations:

1. Creates a copy of the original dataset.
2. Detects and removes exact duplicate records.
3. Standardizes common category spellings.
4. Converts numeric text into numeric values.
5. Checks invalid and out-of-range values.
6. Preserves missing values instead of inventing replacements.

**Important:** The original dataset remains unchanged.

```python
cleaned_df, cleaning_stats = an.clean_data(raw)

print('Original Rows:', len(raw))
print('Cleaned Rows:', len(cleaned_df))
print('Columns:', len(cleaned_df.columns))
print('Duplicate Rows Removed:', cleaning_stats['exact_duplicate_rows_removed'])
print('Missing Cells Before:', cleaning_stats['missing_cells_before'])
print('Missing Cells After:', cleaning_stats['missing_cells_after'])

display(cleaned_df.head())
```

## 5. Saving the Cleaned Dataset

**Purpose:** Save the processed dataset separately from the original file.

`to_csv()` exports the DataFrame as a CSV file.
`index=False` prevents Pandas from adding an unnecessary index column.

```python
cleaned_path = an.PROCESSED_DIR / 'cleaned_dataset.csv'
cleaned_df.to_csv(cleaned_path, index=False)

print('Cleaned Dataset Saved Successfully!')
print('Location:', cleaned_path)
```

## 6. Descriptive Statistical Analysis

**Purpose:** Calculate important statistical measurements.

- Mean: Average value.
- Median: Middle value.
- Mode: Most frequently occurring value.
- Standard deviation: Measures data variation.
- Variance: Squared measure of variation.
- Minimum: Smallest value.
- Maximum: Largest value.

These measurements help understand the distribution of numerical variables.

```python
descriptive, missing_summary = an.descriptive_tables(cleaned_df)

print('Numerical Descriptive Statistics')
display(descriptive)

print('Missing Value Summary')
display(missing_summary)
```

## 7. Categorical Frequency Distribution

**Purpose:** Count the frequency of each category.

For example, we can identify the most frequently used AI tools, user types and AI purposes.

Frequency distributions help identify popular categories and usage patterns.

```python
categorical_data = []

for col in cleaned_df.select_dtypes(exclude=np.number).columns:
    if col == 'User_ID':
        continue
    counts = cleaned_df[col].value_counts(dropna=False)
    for category, count in counts.items():
        categorical_data.append({
            'Column': col,
            'Category': str(category),
            'Frequency': count
        })

frequency_df = pd.DataFrame(categorical_data)
display(frequency_df.head(20))
```

## 8. Outlier Detection Using IQR

**Purpose:** Identify unusually high or low numerical observations.

IQR means Interquartile Range.

Formula:

IQR = Q3 - Q1

Lower Limit = Q1 - 1.5 × IQR

Upper Limit = Q3 + 1.5 × IQR

Values outside these limits are flagged as potential outliers. They are not automatically deleted.

```python
outlier_results = an.outlier_table(cleaned_df)

print('IQR Outlier Analysis')
display(outlier_results)
```

## 9. Correlation Analysis

**Purpose:** Identify relationships between numerical variables.

This project uses Spearman correlation.

Correlation coefficient interpretation:

- +1: Perfect positive relationship.
- 0: No monotonic relationship.
- -1: Perfect negative relationship.

A positive relationship means both variables tend to increase together. Correlation does not establish causation.

```python
correlation_matrix, correlation_results = an.correlation_analysis(cleaned_df)

print('Spearman Correlation Matrix')
display(correlation_matrix)

print('Correlation Test Results')
display(correlation_results)
```

## 10. Correlation Heatmap

**Purpose:** Visualize relationships between numerical variables.

A heatmap uses colors and numerical values to show the strength and direction of correlations.

Values closer to +1 indicate positive associations, while values closer to -1 indicate negative associations.

```python
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, fmt='.2f', cmap='vlag', center=0)
plt.title('Spearman Correlation Heatmap')
plt.tight_layout()
plt.show()
```

## 11. Hypothesis Testing

**Purpose:** Compare students and professionals using the Mann–Whitney U test.

The project compares:

1. Productivity scores.
2. Satisfaction scores.

Null hypothesis (H0): The two groups have the same distribution.

Alternative hypothesis (H1): The two groups have different distributions.

A p-value below 0.05 is commonly treated as statistically significant. The effect size also helps understand the magnitude and direction of the difference.

```python
hypothesis_results = an.hypothesis_tests(cleaned_df)

print('Mann-Whitney U Test Results')
display(hypothesis_results)
```
